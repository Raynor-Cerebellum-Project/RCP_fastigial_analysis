```python
from tqdm import tqdm
import h5py
import gc
from pathlib import Path
import numpy as np

def generate_template(
    chn_data: np.ndarray,  # (NSTIM, interpulse_len)
    template_mode: str,
    repeat_boundaries: np.ndarray,  # 0-based block boundary indices
    window_size: int = 3,
    pca_k: int = 3,
) -> np.ndarray:
    """Generate pulse-by-pulse artifact templates."""
    NSTIM, interpulse_len = chn_data.shape
    template = np.zeros_like(chn_data)
    w = np.linspace(0, 1, interpulse_len)

    # Build is_first_pulse mask
    is_first_pulse = np.zeros(NSTIM, dtype=bool)
    for b in range(len(repeat_boundaries) - 1):
        is_first_pulse[repeat_boundaries[b]] = True

    if template_mode == "local":
        for i in range(NSTIM):
            if is_first_pulse[i]:
                prev_first_idxs = np.where(is_first_pulse[:i])[0]
                if len(prev_first_idxs) >= 1:
                    n_use = min(3, len(prev_first_idxs))
                    ref_idxs = prev_first_idxs[-n_use:]
                    base = chn_data[ref_idxs].mean(axis=0)
                    t = base - base[0]
                    template[i] = t + w * (-t[-1])
                continue
            count = 0
            j = i - 1
            prev_pulses = []
            while j >= 0 and count < window_size:
                if not is_first_pulse[j]:
                    prev_pulses.append(chn_data[j])
                    count += 1
                j -= 1
            if count > 0:
                base = np.stack(prev_pulses).mean(axis=0)
            else:
                base = np.zeros(interpulse_len)
            t = base - base[0]
            template[i] = t + w * (-t[-1])

    elif template_mode == "pca":
        from sklearn.decomposition import PCA
        num_blocks = len(repeat_boundaries) - 1
        for b in range(num_blocks):
            i_start = repeat_boundaries[b]
            i_end   = repeat_boundaries[b + 1]

            # First pulse correction
            prev_first_idxs = np.where(is_first_pulse[:i_start])[0]
            if len(prev_first_idxs) > 0:
                n_use = min(3, len(prev_first_idxs))
                base = chn_data[prev_first_idxs[-n_use:]].mean(axis=0)
            else:
                base = np.zeros(interpulse_len)
            t = base - base[0]
            template[i_start] = t + w * (-t[-1])

            # PCA on rest of block
            if i_end > i_start + 1:
                pca_indices = np.arange(i_start + 1, i_end)
                X = chn_data[pca_indices]
                X_centered = X - X.mean(axis=1, keepdims=True)
                k = min(pca_k, X_centered.shape[0], X_centered.shape[1])
                pca = PCA(n_components=k)
                score = pca.fit_transform(X_centered)
                recon = pca.inverse_transform(score) + X.mean(axis=1, keepdims=True)
                t = recon - recon[:, [0]]
                drift = w[np.newaxis, :] * (-t[:, [-1]])
                template[pca_indices] = t + drift

    return template


def template_subtraction(
    amplifier_data: np.ndarray,
    trigs: np.ndarray,
    params: dict,
    template_mode: str,
    repeat_boundaries: np.ndarray,
    num_repeats: int,
    num_pulse: int,
) -> tuple[np.ndarray, np.ndarray]:
    """Artifact removal via template subtraction + drift correction.

    Returns:
        amplifier_data: corrected trace, shape (n_samples,)
        template: per-pulse template that was subtracted, shape (NSTIM, interpulse_len)
    """
    amplifier_data = amplifier_data.astype(np.float32, copy=True)

    template_leeway   = params["template_leeway"]
    stim_neural_delay = params["stim_neural_delay"]
    window_size       = params["movmean_window"]
    pca_k             = params["pca_k"]
    NSTIM             = trigs.shape[0]

    trigs_beg = trigs[:, 0]
    trigs_end = trigs[:, 1] + stim_neural_delay

    interpulse_len = int(trigs_end[0] - trigs_beg[0] + 1) + 2 * template_leeway
    chn_data = np.full((NSTIM, interpulse_len), np.nan, dtype=np.float32)

    for i in range(NSTIM):
        seg_start = trigs_beg[i] - template_leeway
        seg_end   = trigs_end[i] + template_leeway
        if seg_start < 0 or seg_end >= len(amplifier_data):
            continue
        chn_data[i] = amplifier_data[seg_start:seg_end + 1]

    block_starts = np.array(repeat_boundaries[:-1], dtype=int)
    block_ends   = np.array(repeat_boundaries[1:],  dtype=int)
    boundaries   = np.concatenate([block_starts, [block_ends[-1]]])

    template = generate_template(chn_data, template_mode, boundaries, window_size, pca_k)

    for i in range(NSTIM):
        seg_start = trigs_beg[i] - template_leeway
        seg_end   = trigs_end[i] + template_leeway
        if seg_start < 0 or seg_end >= len(amplifier_data):
            continue
        amplifier_data[seg_start:seg_end + 1] -= template[i]

    # for i in range(NSTIM - 1):
    #     end_i    = trigs_end[i]
    #     beg_next = trigs_beg[i + 1]
    #     mid1 = int(round((end_i + trigs_beg[i]) / 2))
    #     mid2 = int(round((beg_next + trigs_end[i + 1]) / 2))
    #     if mid1 >= mid2 or mid2 >= len(amplifier_data):
    #         continue
    #     seg_idx = np.arange(mid1, mid2 + 1)
    #     template_end_val = amplifier_data[trigs_end[i]]
    #     n = len(seg_idx)
    #     drift = (np.arange(n, dtype=np.float32) / (n - 1) * (-template_end_val)) if n > 1 else np.zeros(1, dtype=np.float32)
    #     amplifier_data[seg_idx] += drift

    return amplifier_data, template


def template_subtraction_all_parallel(
    neural_data: np.ndarray,
    template_params: dict,
    template_mode: str,
    stim_result,
    n_jobs: int = -1,
    out: np.ndarray | None = None,
) -> tuple[np.ndarray, np.ndarray]:
    """Run template subtraction across all channels, writing results in place to avoid duplicate buffers.

    Returns:
        out: corrected neural data, shape (n_channels, n_samples)
        templates: per-channel, per-pulse templates, shape (n_channels, NSTIM, interpulse_len)
    """
    from joblib import Parallel, delayed

    trigs        = stim_result.trigger_pairs
    block_bounds = stim_result.block_bounds_samples
    n_channels, n_samples = neural_data.shape

    pulse_starts = trigs[:, 0]
    repeat_boundaries = [0]
    for b in range(len(block_bounds)):
        block_end_samp = block_bounds[b, 1]
        in_block = np.where(pulse_starts <= block_end_samp)[0]
        if len(in_block):
            repeat_boundaries.append(int(in_block[-1]) + 1)
    repeat_boundaries = np.array(repeat_boundaries, dtype=int)
    num_repeats = len(repeat_boundaries) - 1
    num_pulse   = trigs.shape[0] / num_repeats

    if out is None:
        out = np.empty_like(neural_data, dtype=np.float32)

    templates = [None] * n_channels

    def process_channel(ch):
        corrected, template = template_subtraction(
            neural_data[ch],
            trigs, template_params, template_mode,
            repeat_boundaries, num_repeats, num_pulse,
        )
        out[ch] = corrected
        templates[ch] = template

    Parallel(n_jobs=n_jobs, prefer="threads")(
        delayed(process_channel)(ch) for ch in tqdm(range(n_channels), desc="artifact correction")
    )
    return out, np.stack(templates).astype(np.float32)
```

```python
# Stitch spike trains + stim into copies of this session's Cal files
import shutil
import h5py

CAL_ROOT = Path(str(PARAMS.data_root)) / Path(str(PARAMS.location)) / "Calibrated"
STIM_PREFIX = "stim_"
STIM_EXCLUDE_KEYS = {"stim_traces", "stim_traces_channels_0based"}
SAMPLE_KEY_HINTS = ("sample", "idx", "index", "trigger", "bounds")

_MATLAB_CLASS = {
    "int64": ("int64", 8), "int32": ("int32", 4), "int16": ("int16", 2),
    "uint64": ("uint64", 8), "uint32": ("uint32", 4), "uint16": ("uint16", 2),
    "uint8": ("uint8", 1), "float64": ("double", None), "float32": ("single", None),
}


def write_matlab_array(group, name, arr):
    """Write a numeric array as a MATLAB-readable dataset. Returns True if written."""
    arr = np.asarray(arr)
    if arr.dtype.kind in "OUS":
        return False
    if arr.dtype == np.bool_:
        arr, cls, decode = arr.astype(np.uint8), "logical", 1
    else:
        cls, decode = _MATLAB_CLASS.get(arr.dtype.name, ("double", None))
        if arr.dtype.name not in _MATLAB_CLASS:
            arr = arr.astype(np.float64)

    if arr.ndim == 0:
        arr = arr.reshape(1, 1)
    elif arr.ndim == 1:
        arr = arr.reshape(1, -1)
    else:
        arr = arr.T  # h5py is transposed relative to MATLAB

    if name in group:
        del group[name]
    dset = group.create_dataset(name, data=np.ascontiguousarray(arr))
    dset.attrs["MATLAB_class"] = np.bytes_(cls)
    if decode is not None:
        dset.attrs["MATLAB_int_decode"] = np.int64(decode)
    return True


def align_unit_to_intan(unit_samples, intan_idx):
    """Map a unit's session-relative spike samples onto positions within Intan_idx."""
    _, idx_in_rec, _ = np.intersect1d(intan_idx, np.asarray(unit_samples, np.int64),
                                      return_indices=True)
    return np.sort(idx_in_rec)


def samples_to_intan_pos(samples, intan_idx):
    """Map session-relative samples onto nearest positions within Intan_idx.

    Stim block edges need not appear in Intan_idx, so this snaps to nearest rather
    than intersecting. Samples outside the Cal file's span map to -1. Shape is
    preserved, so an (n, 2) bounds matrix stays (n, 2).
    """
    samples = np.asarray(samples, np.int64)
    intan_idx = np.asarray(intan_idx, np.int64)
    order = np.argsort(intan_idx, kind="stable")
    ordered = intan_idx[order]

    flat = samples.ravel()
    ins = np.clip(np.searchsorted(ordered, flat), 1, ordered.size - 1)
    left, right = ordered[ins - 1], ordered[ins]
    nearest = np.where(np.abs(flat - left) <= np.abs(right - flat), ins - 1, ins)
    pos = order[nearest].astype(np.int64)
    pos[(flat < ordered[0]) | (flat > ordered[-1])] = -1
    return pos.reshape(samples.shape)


def build_stim_fields(stim_arrays, intan_idx, prefix=STIM_PREFIX):
    """Flatten stim arrays into MATLAB-writable fields, with Intan_idx position twins."""
    fields = {}
    for key, arr in stim_arrays.items():
        if key in STIM_EXCLUDE_KEYS:
            continue
        arr = np.asarray(arr)
        if arr.dtype.kind in "OUS":
            continue
        fields[f"{prefix}{key}"] = arr
        if arr.dtype.kind in "iu" and any(h in key.lower() for h in SAMPLE_KEY_HINTS):
            fields[f"{prefix}{key}_pos"] = samples_to_intan_pos(arr, intan_idx)
    return fields

def stitch_cal_file(cal_path, pos_by_uid, extra_fields=None):
    """Copy a v7.3 Cal file and add unit positions plus any extra numeric fields."""
    out_path = cal_path.with_name(cal_path.stem + "_stitched" + cal_path.suffix)
    shutil.copyfile(str(cal_path), str(out_path))

    vlen = h5py.special_dtype(vlen=np.uint8)
    with h5py.File(str(out_path), "r+") as f:
        data = f["Data"]
        if "MATLAB_fields" in data.attrs:
            existing = [bytes(x).decode("ascii") for x in data.attrs["MATLAB_fields"]]
        else:
            existing = list(data.keys())

        n_units, n_extra, new_names = 0, 0, []
        for uid, pos in pos_by_uid.items():
            if pos is None or pos.size == 0:
                continue
            name = f"unit_{uid}"
            if write_matlab_array(data, name, np.asarray(pos, np.int64)):
                if name not in existing and name not in new_names:
                    new_names.append(name)
                n_units += 1

        for name, arr in (extra_fields or {}).items():
            if write_matlab_array(data, name, arr):
                if name not in existing and name not in new_names:
                    new_names.append(name)
                n_extra += 1

        all_fields = existing + new_names
        field_arr = np.empty(len(all_fields), dtype=vlen)
        for i, nm in enumerate(all_fields):
            field_arr[i] = np.frombuffer(nm.encode("ascii"), dtype=np.uint8)
        if "MATLAB_fields" in data.attrs:
            del data.attrs["MATLAB_fields"]
        data.attrs.create("MATLAB_fields", field_arr)

    return out_path, n_units, n_extra


# Locate this session's Cal dir: session i (0-based) -> IntanFile_{i+1}
all_sessions = [s.name for s in rcp.list_intan_sessions(INTAN_ROOT)]
sess_idx = all_sessions.index(sess_name)
cal_dir = CAL_ROOT / f"IntanFile_{sess_idx + 1}"
print(f"{sess_name} is session index {sess_idx} -> {cal_dir.name}")

cal_files = sorted(
    p for p in cal_dir.glob("*_Cal_stim.mat")
    if not p.name.startswith("._") and "_stitched" not in p.stem
)
if not cal_files:
    raise FileNotFoundError(f"no _Cal_stim files in {cal_dir}")
print("stitching:", [p.name for p in cal_files])

for cal_path in cal_files:
    with open(cal_path, "rb") as fh:
        if fh.read(8) != b"\x89HDF\r\n\x1a\n":
            raise OSError(f"{cal_path.name} is not v7.3 — re-save with -v7.3 in MATLAB")

    with h5py.File(str(cal_path), "r") as f:
        intan_idx = np.asarray(f["Data"]["Intan_idx"]).ravel().astype(np.int64)

    pos_by_uid = {int(uid): align_unit_to_intan(st, intan_idx)
                  for uid, st in spike_trains.items()}
    n_nonempty = sum(1 for v in pos_by_uid.values() if v.size)
    stim_fields = build_stim_fields(stim_arrays, intan_idx)

    bb = stim_fields.get(f"{STIM_PREFIX}block_bounds_samples_pos")
    in_span = ""
    if bb is not None and bb.size:
        n_in = int(np.sum(np.all(bb >= 0, axis=1))) if bb.ndim == 2 else int(np.sum(bb >= 0))
        in_span = f", {n_in}/{bb.shape[0]} stim blocks in span"

    out_path, n_units, n_extra = stitch_cal_file(cal_path, pos_by_uid, stim_fields)
    print(f"{cal_path.name}: {n_units}/{len(pos_by_uid)} units "
          f"({n_nonempty} with spikes in span), {n_extra} stim fields{in_span} "
          f"-> {out_path.name}")
```
